# How to check fourcc codec of a video mp4

```python
import cv2

# Open the video file for reading
# FourCC Codec (String): hevc; FourCC Codec (Integer): 1668703592
video_path = '01_001.mp4'  

# cv2.VideoWriter_fourcc(*'mp4v') FourCC Codec (String): FMP4
#video_path = 'output.mp4' 

# cv2.VideoWriter_fourcc(*'xvid') FourCC Codec (String): FMP4
#video_path = 'output.avi' 

video_capture = cv2.VideoCapture(video_path)

# Get the FourCC codec of the video
fourcc = int(video_capture.get(cv2.CAP_PROP_FOURCC))

# Convert the FourCC code to a string
fourcc_str = chr(fourcc & 255) + chr((fourcc >> 8) & 255) + chr((fourcc >> 16) & 255) + chr((fourcc >> 24) & 255)

# Release the video capture object
video_capture.release()

# Print the FourCC codec as a string and as an integer
print(f"FourCC Codec (String): {fourcc_str}")
print(f"FourCC Codec (Integer): {fourcc}")
```

# How to show a sequence of frames as a video on jupyterhub

```python
import cv2
from IPython.display import display, HTML
import os

# Path to the directory containing your frames
frames_directory = '/home/dataset/ped2/training/01'

# Get the list of frame files in the directory
frame_files = [f for f in sorted(os.listdir(frames_directory)) if f.endswith('.jpg')]

# Function to display frames as a video
video_width = 640
video_height = 480
def display_frames_as_video(frame_files, frames_directory):
    video_html = '<video width="500" height="300" controls>'
    
    for frame_file in frame_files:
        frame_path = os.path.join(frames_directory, frame_file)
        video_html += f'<source src="{frame_path}" type="video/mp4">'
    
    video_html += '</video>'
    display(HTML(video_html))

# Display the frames as a video
display_frames_as_video(frame_files, frames_directory)
```

```python
import cv2
from IPython.display import display, HTML
import os

# Path to the directory containing your frames
frames_directory = '/home/dataset/ped2/training/01'

# Get the list of frame files in the directory
frame_files = [f for f in sorted(os.listdir(frames_directory)) if f.endswith('.jpg')]

# Function to display frames as a video
video_width = 640
video_height = 480
frame_paths = []

for frame_file in frame_files:
    frame_path = os.path.join(frames_directory, frame_file)
    frame_paths.append(frame_path)

# Create an HTML5 video element
video_html = f"""
<video width="{video_width}" height="{video_height}" controls>
  {''.join(f'<source src="{frame}" type="image/jpeg">' for frame in frame_paths)}
</video>
"""

# Display the video element
HTML(video_html)
#print(video_html)
```

```python
import ipywidgets as widgets
from IPython.display import display
import cv2
import os

def display_video(frames, fps=30):
    # Create a widget to display the video
    out = widgets.Output(layout={'border': '1px solid black'})
    #display(out)
    height, width, _ = frames[0].shape
    
    # Define the codec and create a VideoWriter object
    '''
    Option 1: 'output.mp4' thi ok
    Video Format: MP4 (MPEG-4 Part 14)
    '''
    #fourcc = cv2.VideoWriter_fourcc(*'mp4v') 
    #video_writer = cv2.VideoWriter('output.mp4', fourcc, fps, (width, height))
    
    '''
    Option 2: 'output.avi' thi ok
    Video Format: AVI (Audio Video Interleave)
    '''
    #fourcc = cv2.VideoWriter_fourcc(*'xvid')
    #video_writer = cv2.VideoWriter('output.avi', fourcc, fps, (width, height))

    '''
    Option 3: 'output.mp4'
    Video Format: 
    '''
    fourcc = cv2.VideoWriter_fourcc(*'hevc')
    video_writer = cv2.VideoWriter('output.mp4', fourcc, fps, (width, height))
    
    # Display each frame and write it to the video
    for frame in frames:
        with out:
            #display(widgets.Image(value=cv2.imencode('.jpg', frame)[1].tobytes()))
            video_writer.write(frame)
    
    # Release the VideoWriter
    video_writer.release()

def CV2_imread(file_path, flag= cv2.IMREAD_COLOR):
    # https://www.geeksforgeeks.org/python-opencv-cv2-imread-method/
    # flag = cv2.IMREAD_COLOR or 1 (default)
    # flag = cv2.IMREAD_GRAYSCALE or 0
    # flag = cv2.IMREAD_UNCHANGED or -1 (alpha channel)
    img_BGR_np = cv2.imread(file_path, flag)
    return img_BGR_np


# Path to the directory containing your frames
frames_directory = '/home/dataset/ped2/training/01'

# Get the list of frame files in the directory
frame_files = [f for f in sorted(os.listdir(frames_directory)) if f.endswith('.jpg')]

# 1. Load your frames into a list (frames_list)
frames = []
for frame_file in frame_files:
    frame_path = os.path.join(frames_directory, frame_file)
    img = CV2_imread(frame_path)
    frames.append(img)

# 2. Display the frames as a video
display_video(frames, fps=30)
```
