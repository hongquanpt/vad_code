# Select a list of elements with different probability

```python
# https://stackoverflow.com/questions/4113307/pythonic-way-to-select-list-elements-with-different-probability
import random
import bisect
import collections

def cdf(weights):
    total = sum(weights)
    result = []
    cumsum = 0
    for w in weights:
        cumsum += w
        result.append(cumsum / total)
    return result

def choice(population, weights):
    assert len(population) == len(weights)
    cdf_vals = cdf(weights)
    x = random.random()
    idx = bisect.bisect(cdf_vals, x)
    return population[idx]

```

```python
def test1():
    print("================================")
    weights=[0.01, 0.99]
    population = 'AB'
    counts = collections.defaultdict(int)
    for i in range(310):
        counts[choice(population, weights)] += 1
    print(counts)
    print(len(counts))

def test2():
    print("================================")
    weights=[0.01, 0.99]
    population = ['anomaly', 'normal']
    anomaly_list = []
    normaly_list = []
    for i in range(310):
        result = choice(population, weights)
        if result == 'anomaly':
            anomaly_list.append(i)
        else:
            normaly_list.append(i)
    print('len(anomaly_list) = ', len(anomaly_list))
    print(anomaly_list)
    print('len(normaly_list) = ', len(normaly_list))
    print(normaly_list)
    return anomaly_list, normaly_list
```

```python
test1()
anomaly_list, normaly_list = test2()
```

```python
i = 100
if (i in anomaly_list):
    print(f"{i}: Element exists in list")
else:
    print(f"{i}: Element not exists in list")
```

```python
def get_lists_dict_with_prob(population = ['anomaly', 'normal'], 
                        weights=[0.01, 0.99], num_elements = 310):
    
    assert len(population) == len(weights)
    assert sum(weights) == 1.0
    print('sum = ', sum(weights))
    lists_dict = collections.defaultdict(list)
    
    for i in range(0, num_elements):
        result = choice(population, weights)
        lists_dict[choice(population, weights)].append(i)
      
    print(lists_dict)
    return lists_dict

lists_dict = get_lists_dict_with_prob()
```
