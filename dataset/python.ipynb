# 1. Xuất ra kích thước ảnh luôn là bội số của 2

```python
def make_power_2(n, base=32.0):
    return int(round(n / base) * base)

w, h = [130, 512]
new_size = [make_power_2(w), make_power_2(h)]
print(new_size)
```

# 4. Operator in python
- [+]	Addition	x + y	
- [-]	Subtraction	x - y	
- [*]	Multiplication	x * y	
- [/]	Division	x / y	
- [%]	Modulus	x % y	
- [**]	Exponentiation	x ** y	
- [//]	Floor division	x // y

```python
a = 10
b = 3

print('a/b = ', a/b) # chia
print('a%b = ', a%b) # chia lấy dư
print('a**b = ', a**b) # hàm mũ
print('a//b = ', a//b) # chia lấy phần nguyên 
```

# 6. Flatten, Unflatten

```python
import torch

img_seq = torch.randn(4, 10, 3, 256, 256)
print(f'img_seq = {img_seq.shape}')

b, t, c, h, w = img_seq.shape
print(f'b = {b}, t = {t}, c = {c}, h = {h}, w = {w}')

img_seq_expand = img_seq.view(-1, c, h, w)
print(f'img_seq_expand = {img_seq_expand.shape}')
```

```python
img_seq = torch.randn(1, 5)
a = img_seq.max(1)
print(f'img_seq={img_seq}')
print(f'a={a}')
```
