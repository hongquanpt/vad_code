# How to use sklearn.cluster.KMeans

```python
from sklearn.cluster import KMeans
import numpy as np

# Generate some random data
np.random.seed(0)
X = np.random.randn(100, 2)  # Create a dataset with 100 samples and 2 features

# Create a KMeans instance with the desired number of clusters (k)
kmeans = KMeans(n_clusters=3)
# Fit the KMeans model to your data
kmeans.fit(X)

# An array of cluster labels for each data point
labels = kmeans.labels_  
print(f'labels = {labels}')

# An array of cluster centers
centroids = kmeans.cluster_centers_  
print(f'centroids = {centroids}')
```

# Find maximum and minimum elements in python

```python
'''
To find the 3 maximum and 3 minimum elements and their indices 
in an array of 10 elements in Python is to use the numpy module, 
which provides functions for working with arrays and matrices. 
For example, you can use the following code:
'''
# Import the numpy module
import numpy as np

# Create an array of 10 elements
arr = np.array([12, 45, 1, 98, 78, 36, 23, 54, 76, 34])
print("The elements in arr are:", arr)
#============================
# Find the 3 minimum elements using np.partition
# Find the 3 maximum elements using np.partition
#min_3 = np.partition(arr, 3)[:3]
#max_3 = np.partition(arr, -3)[-3:]
#print("The 3 minimum elements are:", min_3)
#print("The 3 maximum elements are:", max_3)

#============================
# Find the indices of the 3 minimum elements using np.argpartition
min_3_idx = np.argpartition(arr, 3)[:3]

# Find the indices of the 3 maximum elements using np.argpartition
max_3_idx = np.argpartition(arr, -3)[-3:]

# Print the results
print("The indices of the 3 minimum elements are:", min_3_idx)
print("The indices of the 3 maximum elements are:", max_3_idx)
```
