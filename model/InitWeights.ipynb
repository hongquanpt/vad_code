```python
import torch
import torch.nn as nn
import numpy as np


# Define your deep learning model
class MyModel(nn.Module):
    def __init__(self):
        super(MyModel, self).__init__()
        self.fc1 = nn.Linear(64, 128)
        self.relu1 = nn.ReLU()
        self.fc2 = nn.Linear(128, 64)
        self.relu2 = nn.ReLU()
        self.fc3 = nn.Linear(64, 10)

    def forward(self, x):
        x = self.fc1(x)
        x = self.relu1(x)
        x = self.fc2(x)
        x = self.relu2(x)
        x = self.fc3(x)
        return x

def init_weights():
    # Set a random seed for reproducibility
    seed = 42
    torch.manual_seed(seed)
    np.random.seed(seed)

    # Create an instance of your model
    model = MyModel()
    
    # Apply Xavier Normal Initialization to the model's weights
    def xavier_init(m):
        if isinstance(m, nn.Linear):
            nn.init.xavier_normal_(m.weight)
    
    model.apply(xavier_init)
    
    # The model's weights are now initialized with Xavier initialization and the specified seed
    # Print the weights of the model
    for name, param in model.named_parameters():
        if param.requires_grad:
            print(f"Layer: {name}, Size: {param.size()}, Values: \n{param.data}")
```

```python
init_weights()
```
