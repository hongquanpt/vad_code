```python
# importing libraries
import os
import cv2 
from PIL import Image 

def listdir_nohidden(path):
    out = []
    for f in os.listdir(path):
        if not f.startswith('.') and not os.path.isdir(f):
            out.append(f)
    return out
def generate_video(image_folder, video_filename):
    images = [img for img in sorted(listdir_nohidden(image_folder))
              if img.endswith(".jpg") or
                 img.endswith(".jpeg") or
                 img.endswith("png")]
     
    frame = cv2.imread(os.path.join(image_folder, images[0])) # lấy frame đầu làm thông số
  
    height, width, layers = frame.shape  
    fourcc = cv2.VideoWriter_fourcc(*'XVID')
    video = cv2.VideoWriter(video_filename , fourcc, 20.0, (width, height)) 
  
    for image in images: 
        video.write(cv2.imread(os.path.join(image_folder, image))) # ghi từng ảnh ra video
      
    cv2.destroyAllWindows() 
    video.release()
```
