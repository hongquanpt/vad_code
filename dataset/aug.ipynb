# 7. Patch_Cifar100

```python
from dataset.process_dataset import get_cifar100_dataset

cifar_dataset, cifar_dataloader = get_cifar100_dataset(cfg, 'CIFAR100')
cifar_iter = iter(cifar_dataloader)
        
try:
    # Samples the batch
    cifar_img, _ = next(cifar_iter)
except StopIteration:
    # restart the generator if the previous generator is exhausted.
    cifar_iter = iter(cifar_dataloader)
    cifar_img, _ = next(cifar_iter)

print('cifar_img.shape = ', cifar_img.shape) # torch.Size([1, 3, 32, 32])
Plot_image_tensor_op1(cifar_img)
#train_dataset.set_cifar_img(cifar_img.squeeze(0)) # for get_clip_patch_cifar_()
train_dataset.set_cifar_img(cifar_img)
```
