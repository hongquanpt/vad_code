# torch.stack(tensors, dim=0, *, out=None) → Tensor

- Concatenates a sequence of tensors along a new dimension.
- All tensors need to be of the same size.

```python
# Python 3 program to demonstrate torch.stack() method 
# for two one dimensional tensors 
# importing torch 
import torch 

# creating tensors 
x = torch.randn(3, 5, 5) 
y = torch.randn(3, 5, 5)

# printing above created tensors 
#print("Tensor x:", x) 
#print("Tensor y:", y) 
print("x.shape = ", x.shape)

print("==================================")
# join above tensor using "torch.stack()" 
t = torch.stack((x,y)) 

# print final tensor after join 
#print("join tensors t (default dim=0):\n", t) 
print("t.shape (torch.stack default) = ", t.shape)

print("==================================")
t = torch.stack((x,y), dim = 0) 
#print("join tensors t (dim=0):\n", t) 
print("t.shape (torch.stack dim=0) = ", t.shape)

print("==================================")
t = torch.stack((x,y), dim = 1) 
#print("join tensors t (dim=1):\n", t) 
print("t.shape (torch.stack dim=1) = ", t.shape)
```

# torch.cat(tensors, dim=0, *, out=None) → Tensor
- Concatenates the given sequence of seq tensors in the given dimension. All tensors must either have the same shape (except in the concatenating dimension) or be empty.

- `torch.cat()` can be seen as an inverse operation for `torch.split()` and `torch.chunk()`.

- torch.cat() can be best understood via examples.

```python
# Python 3 program to demonstrate torch.stack() method 
# for two one dimensional tensors 
# importing torch 
import torch 

# creating tensors 
x = torch.randn(3, 5, 5) 
y = torch.randn(3, 5, 5)

# printing above created tensors 
#print("Tensor x:", x) 
#print("Tensor y:", y) 
print("x.shape = ", x.shape)

print("==================================")
# join above tensor using "torch.stack()" 
t = torch.cat((x,y)) 

# print final tensor after join 
#print("join tensors t (default dim=0):\n", t) 
print("t.shape (torch.cat default) = ", t.shape)

print("==================================")
t = torch.cat((x,y), dim = 0) 
#print("join tensors t (dim=0):\n", t) 
print("t.shape (torch.cat dim=0) = ", t.shape)

print("==================================")
t = torch.cat((x,y), dim = 1) 
#print("join tensors t (dim=1):\n", t) 
print("t.shape (torch.cat dim=1) = ", t.shape)
```

# torch.split(tensor, split_size_or_sections, dim=0)
- Splits the tensor into chunks. Each chunk is a view of the original tensor.

- If `split_size_or_sections` is an integer type, then tensor will be split into equally sized chunks (if possible). Last chunk will be smaller if the tensor size along the given dimension dim is not divisible by split_size.

- If `split_size_or_sections` is a list, then tensor will be split into len(split_size_or_sections) chunks with sizes in dim according to split_size_or_sections.

# torch.chunk(input, chunks, dim=0) → List of Tensors
- Attempts to split a tensor into the specified number of chunks. Each chunk is a view of the input tensor.
